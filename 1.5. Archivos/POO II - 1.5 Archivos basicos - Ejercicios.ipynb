## 1. Abrir un archivo con `with open(...)`

Para trabajar con un archivo se usa `open(nombre, modo)`. Escribirlo dentro de un bloque `with` hace que Python **cierre el archivo automáticamente** al terminar, como el *try-with-resources* de Java.

| Modo | Significa | Si el archivo ya existe | Si no existe |
|---|---|---|---|
| `"w"` | escribir (*write*) | **borra** su contenido | lo crea |
| `"r"` | leer (*read*) | lo lee | error `FileNotFoundError` |
| `"a"` | agregar (*append*) | escribe al final | lo crea |

Siempre agregue `encoding="utf-8"` para que las tildes y la ñ se guarden bien.

## 4. Leer línea por línea
Un archivo se puede recorrer con `for`, igual que una lista. Cada línea trae su `\n` al final; `strip()` lo quita.

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    numero = 1
    for linea in archivo:
        print(numero, "→", linea.strip())
        numero += 1
````

## 5. Agregar sin borrar (modo `"a"`)
Con `"w"` se pierde lo anterior; con `"a"` lo nuevo se añade al final.

````python
with open("saludo.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Línea agregada después\n")

with open("saludo.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

## 6. Guardar datos separados por comas
Cada línea es un registro y cada dato se separa con una coma. Para escribir se unen los datos con `f-string`; para leer se separan con `split(",")`.

````python
estudiantes = [("Ana", 9.5), ("Luis", 7.0), ("María", 8.25),("Rene",9.5),("Pedro",10.0)]

with open("notas.txt", "w", encoding="utf-8") as archivo:
    for nombre, nota in estudiantes:
        archivo.write(f"{nombre},{nota}\n")
contador =0
suma=0
with open("notas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        contador += 1
        datos = linea.strip().split(",")
        nombre = datos[0]
        nota = float(datos[1])
        suma += nota# lo que se lee de un archivo siempre es texto
        print(f"{nombre:<8} {nota:>5.2f}")
promedio =suma/contador
print("Promedio es :",promedio)
````

> **Importante:** todo lo que se lee de un archivo es texto (`str`). Para hacer cálculos hay que convertirlo con `int()` o `float()`.

## 7. ¿Y si el archivo no existe?
Leer un archivo que no existe produce `FileNotFoundError` y el programa se detiene. Con `try / except` se muestra un mensaje y el programa continúa.

````python
try:
    with open("no_existe.txt", "r", encoding="utf-8") as archivo:
        print(archivo.read())
except FileNotFoundError:
    print("El archivo no existe, revise el nombre")

print("El programa sigue funcionando")
````

## Resumen

| Quiero… | Código |
|---|---|
| Crear o reemplazar | `with open("a.txt", "w", encoding="utf-8") as f:` `f.write("texto\n")` |
| Leer todo | `f.read()` |
| Leer línea por línea | `for linea in f:` `linea.strip()` |
| Agregar al final | `open("a.txt", "a", encoding="utf-8")` |
| Separar datos | `linea.strip().split(",")` |
| Archivo que no existe | `try:` … `except FileNotFoundError:` |

### Ejercicio 8. Promedio del curso
Lea `calificaciones.txt` y calcule el **promedio** de las notas. Muestre también cuántos estudiantes **aprobaron** (nota mayor o igual a 7).

> 💡 **Pista:** Recuerde convertir la nota con `int()`.

````python
suma_notas = 0
total_estudiantes = 0
aprobados = 0

with open("calificaciones.txt", "r", encoding="utf-8") as archivo:
    for estudiante in archivo:
        nombre, nota_texto = estudiante.strip().split(",")
        nota = int(nota_texto)
        
        suma_notas += nota
        total_estudiantes += 1
        
        if nota >= 7:
            aprobados += 1

promedio = suma_notas / total_estudiantes

print(f"Promedio del curso: {promedio:.2f}")
print(f"Estudiantes aprobados: {aprobados}")
````

````python

with open("calificaciones.txt", "r", encoding="utf-8") as entrada, \
    open("aprobados.txt", "w", encoding="utf-8") as salida:
    
    for estudiante in entrada:
        nombre, nota_texto = estudiante.strip().split(",")
        if int(nota_texto) >= 7:
            salida.write(estudiante)


print("--- Estudiantes aprobados ---")
with open("aprobados.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

```

   BIBLIOTECA DIGITAL TEC AZUAY

2. Mostrar libros
3. Buscar libro
4. Prestar libro
```

El menú se repite con un `while` hasta que el usuario elija la opción 8.

## Funcionalidades
1. **Registrar libro:** pide el título y el autor, asigna el código siguiente de forma automática y queda disponible.
2. **Mostrar libros:** presenta una tabla alineada con código, título, autor y estado (Disponible o Prestado).
3. **Buscar libro:** pide una palabra y muestra los libros cuyo título la contenga, sin distinguir mayúsculas de minúsculas. Si no encuentra ninguno, lo indica.
4. **Prestar libro:** pide el código del libro y el nombre del estudiante. Solo se presta si el libro existe y está disponible. Registra el préstamo con un número consecutivo y la fecha.
5. **Devolver libro:** pide el código del libro. Si estaba prestado, vuelve a quedar disponible y el préstamo se marca como devuelto.
6. **Mostrar préstamos:** lista los préstamos con número, libro, estudiante, fecha y estado (Activo o Devuelto).
7. **Reporte general:** muestra el total de libros, los disponibles, los prestados, los préstamos activos y el porcentaje de libros prestados.
8. **Salir:** guarda los datos y muestra un mensaje de despedida.

````python
from datetime import date

ARCHIVO_LIBROS = "libros.txt"
ARCHIVO_PRESTAMOS = "prestamos.txt"


# --- FUNCIONES DE ARCHIVOS ---

def cargar_libros():
    """Lee libros.txt. Si no existe, lo crea con los 3 libros iniciales."""
    libros = []
    try:
        with open(ARCHIVO_LIBROS, "r", encoding="utf-8") as archivo:
            for linea in archivo:
                linea = linea.strip()
                if linea:
                    codigo, titulo, autor, estado = linea.split(",")
                    libros.append({
                        "codigo": int(codigo),
                        "titulo": titulo,
                        "autor": autor,
                        "estado": estado
                    })
    except FileNotFoundError:
        libros = [
            {"codigo": 1, "titulo": "Python para todos", "autor": "Charles Severance", "estado": "Disponible"},
            {"codigo": 2, "titulo": "Aprende Java", "autor": "Ana Torres", "estado": "Disponible"},
            {"codigo": 3, "titulo": "Bases de datos", "autor": "Carlos Mora", "estado": "Disponible"}
        ]
        guardar_libros(libros)
    return libros


def guardar_libros(libros):
    """Guarda la lista de libros en libros.txt separados por comas."""
    with open(ARCHIVO_LIBROS, "w", encoding="utf-8") as archivo:
        for libro in libros:
            archivo.write(f"{libro['codigo']},{libro['titulo']},{libro['autor']},{libro['estado']}\n")


def cargar_prestamos():
    """Lee prestamos.txt. Si no existe, crea el archivo vacío y retorna lista vacía."""
    prestamos = []
    try:
        with open(ARCHIVO_PRESTAMOS, "r", encoding="utf-8") as archivo:
            for linea in archivo:
                linea = linea.strip()
                if linea:
                    num, cod_libro, estudiante, fecha, estado = linea.split(",")
                    prestamos.append({
                        "numero": int(num),
                        "codigo_libro": int(cod_libro),
                        "estudiante": estudiante,
                        "fecha": fecha,
                        "estado": estado
                    })
    except FileNotFoundError:
        guardar_prestamos(prestamos)
    return prestamos


def guardar_prestamos(prestamos):
    """Guarda la lista de préstamos en prestamos.txt separados por comas."""
    with open(ARCHIVO_PRESTAMOS, "w", encoding="utf-8") as archivo:
        for p in prestamos:
            archivo.write(f"{p['numero']},{p['codigo_libro']},{p['estudiante']},{p['fecha']},{p['estado']}\n")


# --- FUNCIONES DE APOYO ---

def buscar_libro_por_codigo(libros, codigo):
    """Busca y retorna el diccionario del libro según su código, o None si no existe."""
    for libro in libros:
        if libro["codigo"] == codigo:
            return libro
    return None


def leer_entero(mensaje):
    """Solicita un número entero controlando ValueError para evitar caídas."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("❌ Error: Debe ingresar un número entero válido.")


def leer_texto_no_vacio(mensaje):
    """Solicita un texto y valida que no quede vacío ni tenga comas que dañen el .txt."""
    while True:
        texto = input(mensaje).strip()
        if not texto:
            print("❌ Error: Este campo no puede quedar vacío.")
        elif "," in texto:
            print("❌ Error: No use comas (,) en el texto para no alterar el archivo.")
        else:
            return texto


# --- FUNCIONES DE LAS OPCIONES DEL MENÚ ---

def registrar_libro(libros):
    """Opción 1: Registra un nuevo libro con código automático y lo guarda."""
    print("\n--- REGISTRAR NUEVO LIBRO ---")
    titulo = leer_texto_no_vacio("Título del libro: ")
    autor = leer_texto_no_vacio("Autor del libro: ")

    nuevo_codigo = max([l["codigo"] for l in libros], default=0) + 1
    nuevo_libro = {
        "codigo": nuevo_codigo,
        "titulo": titulo,
        "autor": autor,
        "estado": "Disponible"
    }
    libros.append(nuevo_libro)
    guardar_libros(libros)
    print(f"✅ Libro '{titulo}' registrado exitosamente con el código {nuevo_codigo}.")
    return libros


def mostrar_libros(libros):
    """Opción 2: Muestra todos los libros en una tabla alineada."""
    print("\n--- CATÁLOGO DE LIBROS ---")
    if not libros:
        print("No hay libros registrados.")
        return 0

    encabezado = f"{'Código':<8} | {'Título':<28} | {'Autor':<22} | {'Estado':<12}"
    print(encabezado)
    print("-" * len(encabezado))
    for libro in libros:
        print(f"{libro['codigo']:<8} | {libro['titulo']:<28} | {libro['autor']:<22} | {libro['estado']:<12}")
    return len(libros)


def buscar_libro(libros):
    """Opción 3: Busca libros por coincidencia parcial en el título (sin importar mayúsculas)."""
    print("\n--- BUSCAR LIBRO ---")
    palabra = leer_texto_no_vacio("Ingrese palabra clave del título: ").lower()
    encontrados = [l for l in libros if palabra in l["titulo"].lower()]

    if not encontrados:
        print(f"No se encontraron libros que contengan '{palabra}'.")
        return []

    encabezado = f"{'Código':<8} | {'Título':<28} | {'Autor':<22} | {'Estado':<12}"
    print(encabezado)
    print("-" * len(encabezado))
    for libro in encontrados:
        print(f"{libro['codigo']:<8} | {libro['titulo']:<28} | {libro['autor']:<22} | {libro['estado']:<12}")
    return encontrados


def prestar_libro(libros, prestamos):
    """Opción 4: Registra el préstamo de un libro disponible."""
    print("\n--- PRESTAR LIBRO ---")
    codigo = leer_entero("Ingrese el código del libro a prestar: ")
    libro = buscar_libro_por_codigo(libros, codigo)

    if libro is None:
        print(f"❌ Error: No existe ningún libro con el código {codigo}.")
        return False

    if libro["estado"] != "Disponible":
        print(f"❌ Error: El libro '{libro['titulo']}' ya se encuentra Prestado.")
        return False

    estudiante = leer_texto_no_vacio("Nombre del estudiante: ")
    numero_prestamo = len(prestamos) + 1
    fecha_actual = date.today().isoformat()  # Formato YYYY-MM-DD (ej. 2026-11-17)

    nuevo_prestamo = {
        "numero": numero_prestamo,
        "codigo_libro": codigo,
        "estudiante": estudiante,
        "fecha": fecha_actual,
        "estado": "Activo"
    }

    libro["estado"] = "Prestado"
    prestamos.append(nuevo_prestamo)
    guardar_libros(libros)
    guardar_prestamos(prestamos)

    print(f"✅ Préstamo #{numero_prestamo} registrado: '{libro['titulo']}' prestado a {estudiante}.")
    return True


def devolver_libro(libros, prestamos):
    """Opción 5: Marca un libro prestado como Disponible y cierra su préstamo activo."""
    print("\n--- DEVOLVER LIBRO ---")
    codigo = leer_entero("Ingrese el código del libro a devolver: ")
    libro = buscar_libro_por_codigo(libros, codigo)

    if libro is None:
        print(f"❌ Error: No existe ningún libro con el código {codigo}.")
        return False

    if libro["estado"] == "Disponible":
        print(f"❌ Error: El libro '{libro['titulo']}' no está prestado actualmente.")
        return False

    libro["estado"] = "Disponible"
    for p in prestamos:
        if p["codigo_libro"] == codigo and p["estado"] == "Activo":
            p["estado"] = "Devuelto"
            break

    guardar_libros(libros)
    guardar_prestamos(prestamos)
    print(f"✅ Libro '{libro['titulo']}' devuelto correctamente. Ahora está Disponible.")
    return True


def mostrar_prestamos(libros, prestamos):
    """Opción 6: Lista todos los préstamos con el título del libro correspondiente."""
    print("\n--- HISTORIAL DE PRÉSTAMOS ---")
    if not prestamos:
        print("Aún no se han registrado préstamos.")
        return 0

    encabezado = f"{'N°':<5} | {'Libro':<26} | {'Estudiante':<22} | {'Fecha':<12} | {'Estado':<10}"
    print(encabezado)
    print("-" * len(encabezado))

    for p in prestamos:
        libro = buscar_libro_por_codigo(libros, p["codigo_libro"])
        titulo_libro = libro["titulo"] if libro else f"Código {p['codigo_libro']}"
        print(f"{p['numero']:<5} | {titulo_libro:<26} | {p['estudiante']:<22} | {p['fecha']:<12} | {p['estado']:<10}")
    return len(prestamos)


def reporte_general(libros, prestamos):
    """Opción 7: Calcula y muestra estadísticas generales de la biblioteca."""
    print("\n--- REPORTE GENERAL DE BIBLIOTECA ---")
    total_libros = len(libros)
    disponibles = sum(1 for l in libros if l["estado"] == "Disponible")
    prestados = sum(1 for l in libros if l["estado"] == "Prestado")
    prestamos_activos = sum(1 for p in prestamos if p["estado"] == "Activo")
    porcentaje_prestados = (prestados / total_libros * 100) if total_libros > 0 else 0.0

    print(f"Total de libros registrados : {total_libros}")
    print(f"Libros disponibles          : {disponibles}")
    print(f"Libros prestados            : {prestados}")
    print(f"Préstamos activos           : {prestamos_activos}")
    print(f"Porcentaje de préstamos     : {porcentaje_prestados:.2f}%")

    return {
        "total": total_libros,
        "disponibles": disponibles,
        "prestados": prestados,
        "activos": prestamos_activos,
        "porcentaje": porcentaje_prestados
    }


def salir_sistema(libros, prestamos):
    """Opción 8: Guarda ambos archivos y retorna False para finalizar el menú."""
    guardar_libros(libros)
    guardar_prestamos(prestamos)
    print("\n💾 Datos guardados correctamente en libros.txt y prestamos.txt.")
    print("👋 ¡Gracias por usar la Biblioteca Digital TEC Azuay! Hasta pronto.")
    return False


# --- PROGRAMA PRINCIPAL ---

def main():
    libros = cargar_libros()
    prestamos = cargar_prestamos()
    ejecutando = True

    while ejecutando:
        print("""
==================================
   BIBLIOTECA DIGITAL TEC AZUAY
==================================
1. Registrar libro
2. Mostrar libros
3. Buscar libro
4. Prestar libro
5. Devolver libro
6. Mostrar préstamos
7. Reporte general
8. Salir""")

        opcion = leer_entero("Seleccione una opción (1-8): ")

        match opcion:
            case 1:
                registrar_libro(libros)
            case 2:
                mostrar_libros(libros)
            case 3:
                buscar_libro(libros)
            case 4:
                prestar_libro(libros, prestamos)
            case 5:
                devolver_libro(libros, prestamos)
            case 6:
                mostrar_prestamos(libros, prestamos)
            case 7:
                reporte_general(libros, prestamos)
            case 8:
                ejecutando = salir_sistema(libros, prestamos)
            case _:
                print("❌ Error: Opción fuera de rango. Elija un número del 1 al 8.")


if __name__ == "__main__":
    main()
````
