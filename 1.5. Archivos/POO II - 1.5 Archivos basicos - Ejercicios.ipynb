## 1. Abrir un archivo con `with open(...)`

Para trabajar con un archivo se usa `open(nombre, modo)`. Escribirlo dentro de un bloque `with` hace que Python **cierre el archivo automáticamente** al terminar, como el *try-with-resources* de Java.

| Modo | Significa | Si el archivo ya existe | Si no existe |
|---|---|---|---|
| `"w"` | escribir (*write*) | **borra** su contenido | lo crea |
| `"r"` | leer (*read*) | lo lee | error `FileNotFoundError` |
| `"a"` | agregar (*append*) | escribe al final | lo crea |

Siempre agregue `encoding="utf-8"` para que las tildes y la ñ se guarden bien.

## 4. Leer línea por línea
Un archivo se puede recorrer con `for`, igual que una lista. Cada línea trae su `\n` al final; `strip()` lo quita.

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    numero = 1
    for linea in archivo:
        print(numero, "→", linea.strip())
        numero += 1
````

````python
with open("persona3.txt", "r", encoding="utf-8") as archivo:
    numero = 0
    for linea in archivo:
        print(numero, "→", linea.strip())
        numero += 1
````

También se pueden obtener todas las líneas en una lista con `readlines()`:

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

print(lineas)
print("El archivo tiene", len(lineas), "líneas")
````

## 5. Agregar sin borrar (modo `"a"`)
Con `"w"` se pierde lo anterior; con `"a"` lo nuevo se añade al final.

````python
with open("saludo.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Línea agregada después\n")

with open("saludo.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

````python
persona6= {"nombre": "Miguel",   "edad": 25, "correo": "migue@gmail.com",   "cedula": "0109988776", "residencia": "Guayaquil", "activo": False}
persona7= {"nombre": "Debora",   "edad": 27, "correo": "debora@gmail.com",   "cedula": "0109988777", "residencia": "Guayaquil", "activo": False}
with open("persona3.txt", "a", encoding="utf8") as archivo:
    archivo.write(f"persona 7: {persona7}\n")
    for clave, valor in persona6.items():
            archivo.write(f"{clave}: {valor}\n")
with open("persona3.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

## 6. Guardar datos separados por comas
Cada línea es un registro y cada dato se separa con una coma. Para escribir se unen los datos con `f-string`; para leer se separan con `split(",")`.

````python
estudiantes = [("Ana", 9.5), ("Luis", 7.0), ("María", 8.25),("Rene",9.5),("Pedro",10.0)]

with open("notas.txt", "w", encoding="utf-8") as archivo:
    for nombre, nota in estudiantes:
        archivo.write(f"{nombre},{nota}\n")
contador =0
suma=0
with open("notas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        contador += 1
        datos = linea.strip().split(",")
        nombre = datos[0]
        nota = float(datos[1])
        suma += nota# lo que se lee de un archivo siempre es texto
        print(f"{nombre:<8} {nota:>5.2f}")
promedio =suma/contador
print("Promedio es :",promedio)
````

> **Importante:** todo lo que se lee de un archivo es texto (`str`). Para hacer cálculos hay que convertirlo con `int()` o `float()`.

## 7. ¿Y si el archivo no existe?
Leer un archivo que no existe produce `FileNotFoundError` y el programa se detiene. Con `try / except` se muestra un mensaje y el programa continúa.

````python
try:
    with open("no_existe.txt", "r", encoding="utf-8") as archivo:
        print(archivo.read())
except FileNotFoundError:
    print("El archivo no existe, revise el nombre")

print("El programa sigue funcionando")
````

## Resumen

| Quiero… | Código |
|---|---|
| Crear o reemplazar | `with open("a.txt", "w", encoding="utf-8") as f:` `f.write("texto\n")` |
| Leer todo | `f.read()` |
| Leer línea por línea | `for linea in f:` `linea.strip()` |
| Agregar al final | `open("a.txt", "a", encoding="utf-8")` |
| Separar datos | `linea.strip().split(",")` |
| Archivo que no existe | `try:` … `except FileNotFoundError:` |

```

   BIBLIOTECA DIGITAL TEC AZUAY

2. Mostrar libros
3. Buscar libro
4. Prestar libro
```

El menú se repite con un `while` hasta que el usuario elija la opción 8.

## Funcionalidades
1. **Registrar libro:** pide el título y el autor, asigna el código siguiente de forma automática y queda disponible.
2. **Mostrar libros:** presenta una tabla alineada con código, título, autor y estado (Disponible o Prestado).
3. **Buscar libro:** pide una palabra y muestra los libros cuyo título la contenga, sin distinguir mayúsculas de minúsculas. Si no encuentra ninguno, lo indica.
4. **Prestar libro:** pide el código del libro y el nombre del estudiante. Solo se presta si el libro existe y está disponible. Registra el préstamo con un número consecutivo y la fecha.
5. **Devolver libro:** pide el código del libro. Si estaba prestado, vuelve a quedar disponible y el préstamo se marca como devuelto.
6. **Mostrar préstamos:** lista los préstamos con número, libro, estudiante, fecha y estado (Activo o Devuelto).
7. **Reporte general:** muestra el total de libros, los disponibles, los prestados, los préstamos activos y el porcentaje de libros prestados.
8. **Salir:** guarda los datos y muestra un mensaje de despedida.
